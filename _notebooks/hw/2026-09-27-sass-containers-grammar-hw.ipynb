### Popcorn Hack: Parade Invite Rescue

**Time:** ~5 minutes

The Poway Scripps Rotary Parade invite is broken. It uses:
- a made-up `page-wrap` class
- inline CSS
- a hand-made flex row
- made-up `box` classes

Your job is to rebuild it using **only OCS container grammar**.

### Your Tasks

1. Replace the `page-wrap` div with one `ocs__container`.
2. Put the content inside an `ocs__card`.
3. Change the title to `ocs__section-title`.
4. Add a one-line `ocs__description`.
5. Replace the flex row and `box` classes with:
   - `ocs__grid`
   - `ocs__grid--standard`
   - `cols-3`
6. Give each schedule item an `ocs__grid-cell`.
7. Give one schedule item the `ocs__grid-cell--accent` class.
8. Remove every inline `style` attribute.
9. Remove every made-up class.

Press **Run** after making your changes.

### Done When

- There is one `ocs__container`.
- There is one `ocs__card`.
- There is one standard 3-column grid.
- There are three `ocs__grid-cell` elements.
- One cell uses `ocs__grid-cell--accent`.
- There are no inline styles.
- There are no made-up classes.

Resize the page to make sure the layout remains responsive.

let myHTML = [
  '<div class="wrap" style="max-width: 1100px; margin: auto;">',

    classList.includes('ocs__table')) +

  line('Has an ocs__callout',
    classList.includes('ocs__callout')) +

outputElement.innerHTML = checks + myHTML;
