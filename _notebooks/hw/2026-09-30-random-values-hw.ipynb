# 3.15 Random Values

## Popcorn Hack 1 — Random SFI Record Number

Use `random.randint()` to randomly generate an SFI record number.

`random.randint(1, 4)` includes both endpoints, so the possible results are:

- 1
- 2
- 3
- 4

A result can repeat because each selection is random.

```python
# CODE_RUNNER: Popcorn Hack 1

import random

# There are four possible SFI record numbers.
record_count = 4

# randint() includes both endpoints.
sfi_record_id = random.randint(1, record_count)

print("SFI record selected for QA:", sfi_record_id)
print("Possible record IDs: 1, 2, 3, 4")
print("Repeated results are allowed because the selection is random.")
```

## Popcorn Hack 2 — Random Structured SFI Part

The backend stores each SFI part as a complete record.

Use `random.choice()` to select one complete record. This keeps the product name, category, and spec number together.

Run the code multiple times and observe how the selected record can change.

```python
# CODE_RUNNER: Popcorn Hack 2

import random

# Each dictionary represents one complete SFI record.
parts = [
    {
        "product_name": "Replacement Flywheels and Clutch Assemblies",
        "category": "Auto Racing",
        "spec_number": "1.1"
    },
    {
        "product_name": "Multiple Disc Clutch Assemblies",
        "category": "Drag Racing",
        "spec_number": "1.2"
    },
    {
        "product_name": "Racing Flywheel Record",
        "category": "Auto Racing",
        "spec_number": "2.1"
    }
]

# Randomly choose one complete record.
selected_part = random.choice(parts)

print("SFI part selected for testing:")
print("Product:", selected_part["product_name"])
print("Category:", selected_part["category"])
print("Spec number:", selected_part["spec_number"])
```

### Homework Hack (independent): SFI Backend QA Simulator

Build a reusable randomized QA simulator for the SFI automotive backend.

The simulator should work with:
- structured SFI part records,
- multiple backend actions,
- existing spec numbers,
- repeated test runs,
- and duplicate-spec handling during creation.

The final runner should clearly show which record and action were selected and what backend-style result occurred.

Copy this starter into your **personal submission notebook** after class. Complete it there, run multiple tests, and include the output. This is a reference block, not an in-page runner:

```python
# CODE_RUNNER: Homework Hack
# Homework Hack - SFI Backend QA Simulator

import random

# Structured SFI automotive records.
parts = [
    {
        "product_name": "Replacement Flywheels",
        "category": "Auto Racing",
        "spec_number": "1.1"
    },
    {
        "product_name": "Multiple Disc Clutch Assemblies",
        "category": "Drag Racing",
        "spec_number": "1.2"
    },
    {
        "product_name": "Racing Flywheel Record",
        "category": "Auto Racing",
        "spec_number": "2.1"
    }
]

# Possible backend actions.
actions = [
    "GET/search",
    "POST/create",
    "PUT/update",
    "DELETE/remove"
]

# Spec numbers that already exist in the backend.
existing_spec_numbers = ["1.1", "2.1"]

# Number of QA tests to run.
test_count = 5


# Run the randomized QA simulator.
for test_number in range(1, test_count + 1):

    # Randomly select one complete SFI record.
    selected_part = random.choice(parts)

    # Randomly select one backend action.
    selected_action = random.choice(actions)

    print(f"\n--- QA Test {test_number} ---")
    print("Selected part:", selected_part["product_name"])
    print("Category:", selected_part["category"])
    print("Spec number:", selected_part["spec_number"])
    print("Selected action:", selected_action)

    # Handle GET/search.
    if selected_action == "GET/search":
        print("Backend result: SUCCESS - SFI record found.")

    # Handle POST/create.
    elif selected_action == "POST/create":

        # Check whether the spec number already exists.
        if selected_part["spec_number"] in existing_spec_numbers:
            print("Backend result: ERROR - duplicate spec number.")
        else:
            print("Backend result: SUCCESS - SFI record created.")

    # Handle PUT/update.
    elif selected_action == "PUT/update":
        print("Backend result: SUCCESS - SFI record updated.")

    # Handle DELETE/remove.
    elif selected_action == "DELETE/remove":
        print("Backend result: SUCCESS - SFI record removed.")
```

## Random Values Summary

`random.randint(1, 4)` can produce 1, 2, 3, or 4 because both endpoints are included.

`random.choice(parts)` selects one complete item from the list.

The final simulator has:

- 3 possible SFI records
- 4 possible backend actions
- 12 possible record/action combinations

Random combinations can repeat across different tests. Repetition is normal behavior for random selection.
