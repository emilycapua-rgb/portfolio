### Popcorn Hack 2: Size the Beakers with `fr`

**Task:** Put all four beakers in one row.

The **Control Beaker** should be **twice as wide** as each of the other three beakers.

**Goal:** Use `fr` values to control the width of each column.

Remember:

- The number of `fr` values determines how many boxes fit in a row.
- A larger `fr` value gives that column more space.
- Four boxes in one row requires four `fr` values.
- The Control Beaker is the last box.

Press **Run** after changing the grid.

{% capture popcorn2 %}
Popcorn Hack 2: change the fr values so all 4 beakers sit in one row and the Control Beaker is twice as wide.
{% endcapture %}

{% capture popcorn2_code %}
// Change the grid-template-columns value, then press Run.

let myHTML = [
  '<div class="ocs__grid ocs__grid--standard" style="grid-template-columns: 1fr 1fr 1fr 2fr;">',
  '  <div class="ocs__grid-cell">Beaker 1</div>',
  '  <div class="ocs__grid-cell">Beaker 2</div>',
  '  <div class="ocs__grid-cell">Beaker 3</div>',
  '  <div class="ocs__grid-cell">Control Beaker</div>',
  '</div>'
].join('');

// ---- You do not need to change anything below this line ----

function line(label, ok) {
  return '<li>' + (ok ? 'PASS  ' : 'FIX  ') + label + '</li>';
}

// Find the grid-template-columns value.
let match = myHTML.match(/grid-template-columns:\s*([^;"]+)/);
let columns = match ? match[1].trim() : '';

let values = columns.split(/\s+/).filter(v => v.length > 0);

let allFr = values.every(v => /^\d+(\.\d+)?fr$/.test(v));

let numbers = values.map(v => parseFloat(v));

let fourColumns = numbers.length === 4;

let controlTwiceAsWide =
  fourColumns &&
  numbers[3] === numbers[0] * 2 &&
  numbers[1] === numbers[0] &&
  numbers[2] === numbers[0];

let checks = '<ul>' +
  line(
    'Uses four fr values',
    fourColumns
  ) +
  line(
    'All column values use fr',
    allFr
  ) +
  line(
    'Control Beaker is twice as wide',
    controlTwiceAsWide
  ) +
  '</ul>';

outputElement.innerHTML = checks + myHTML;
