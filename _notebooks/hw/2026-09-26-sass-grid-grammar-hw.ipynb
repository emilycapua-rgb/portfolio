### Homework Hack

**Task:** Fix the code below. Remove every inline style and made up class and use OCS classes instead.

* **Grid 1:** a standard grid. The title becomes a **header** cell. Beaker 3 is the outlier, so it becomes an **accent** cell.
* **Grid 2:** a **calculator** grid, which is a keypad for typing in beaker readings. The `Avg` key becomes an **accent** cell.

**Goal:** this is the whole lesson in one task. You pick the variant, put a cell class on every box, and use the right cell type for the title and the outlier.

Edit the HTML inside the quotes and press **Run**. The checklist at the top of the output marks each rule **PASS** or **FIX**. Keep going until every line says **PASS**.

// Fix the HTML inside the quotes, then press Run.
let myHTML = [
  '<div class="my-row">',
  '  <div class="title-box" style="width: 100%;">Temperature Trial (°C)</div>',
  '  <div class="sample-card">Beaker 1: 22.4</div>',
  '  <div class="sample-card">Beaker 2: 22.9</div>',
  '  <div class="sample-card highlight">Beaker 3: 31.6</div>',
  '</div>',
  '<div style="display: grid; grid-template-columns: repeat(4, 1fr);">',
  '  <div class="key">7</div>',
  '  <div class="key">8</div>',
  '  <div class="key">9</div>',
  '  <div class="key orange">Avg</div>',
  '</div>'
].join('');

// ---- You do not need to change anything below this line ----
let classList = [];
for (let found of (myHTML.match(/class="[^"]*"/g) || [])) {
  classList = classList.concat(found.slice(7, -1).split(' ').filter(c => c !== ''));
}
let madeUp = classList.filter(c => !c.startsWith('ocs__') && !c.startsWith('cols-'));

function line(label, ok) { return '<li>' + (ok ? 'PASS  ' : 'FIX  ') + label + '</li>'; }
let checks = '<ul>' +
  line('No inline style attributes', !myHTML.includes('style=')) +
  line('No made up classes' + (madeUp.length > 0 ? ' (found: ' + madeUp.join(', ') + ')' : ''), madeUp.length === 0) +
  line('Has an ocs__grid root', classList.includes('ocs__grid')) +
  line('Uses the standard variant for Grid 1', classList.includes('ocs__grid--standard')) +
  line('Uses the calculator variant for Grid 2', classList.includes('ocs__grid--calculator')) +
  line('Every box uses ocs__grid-cell', classList.filter(c => c === 'ocs__grid-cell').length >= 8) +
  line('Title is a header cell', classList.includes('ocs__grid-cell--header')) +
  line('Has accent cells (outlier beaker and Avg key)', classList.filter(c => c === 'ocs__grid-cell--accent').length >= 2) +
  '</ul>';

outputElement.innerHTML = checks + myHTML;
