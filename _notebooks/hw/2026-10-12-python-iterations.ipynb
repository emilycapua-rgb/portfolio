## Popcorn Hack: Wildfire Dispatch Scanner

Complete the code to scan wildfire alert levels.

Requirements:
1. Choose a cutoff for safe alerts.
2. Choose a cutoff for critical alerts.
3. Loop through the `alerts` list.
4. Use `continue` to skip safe alerts.
5. Count action alerts and add their scores to a running total.
6. Use `break` to stop at the first critical alert.
7. Print a final summary.

```python
# CODE_RUNNER: Iterations Popcorn - Wildfire Dispatch Scanner

alerts = [1, 4, 2, 6, 3, 9, 5]

safe_cutoff = 3
critical_cutoff = 8

action_count = 0
safe_count = 0
total_action_score = 0
critical_found = False

for alert in alerts:
    if alert < safe_cutoff:
        safe_count += 1
        print("Safe alert skipped:", alert)
        continue

    action_count += 1
    total_action_score += alert
    print("Dispatch responding to alert:", alert)

    if alert >= critical_cutoff:
        print("Critical SFSRC alert found")
        critical_found = True
        break

print("Safe alerts skipped:", safe_count)
print("Action alerts processed:", action_count)
print("Total action score:", total_action_score)
print("Critical alert found:", critical_found)
```

## Homework Hack: Build a Loop-Based Checker

Create a monitoring system using a theme such as wildfire alerts, server loads, battery levels, quiz scores, or air-quality readings.

Your program must:
1. Include at least 8 values in a list.
2. Define a safe cutoff and a critical cutoff.
3. Use a `for` loop to inspect the values.
4. Use `continue` to skip values below the safe cutoff.
5. Use `break` to stop at the first critical value.
6. Track the checked count, action count, skipped count, and running total.
7. Print a labeled final report.

Run your code and check that it stops when a critical value is found.

```python
# CODE_RUNNER: Iterations Homework - Loop-Based Checker

# Theme: SFSRC wildfire response
wildfire_alerts = [1, 4, 2, 6, 3, 5, 8, 4, 7]

safe_cutoff = 3
critical_cutoff = 8

checked_count = 0
action_count = 0
skipped_count = 0
total_score = 0
critical_found = False

for alert in wildfire_alerts:
    checked_count += 1

    if alert < safe_cutoff:
        skipped_count += 1
        print("Safe alert skipped:", alert)
        continue

    action_count += 1
    total_score += alert
    print("Action required for alert:", alert)

    if alert >= critical_cutoff:
        print("Critical alert found:", alert)
        critical_found = True
        break

print("\n--- Final Monitoring Report ---")
print("Values checked:", checked_count)
print("Values skipped:", skipped_count)
print("Action alerts:", action_count)
print("Total action score:", total_score)
print("Critical alert found:", critical_found)
```
