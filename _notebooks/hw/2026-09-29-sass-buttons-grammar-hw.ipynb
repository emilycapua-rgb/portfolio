```python
{% capture homework1 %}
Homework Hack 1: Build the Adaptive Resource Toolbar using only OCS button classes.
{% endcapture %}

{% capture homework1_code %}
// Refactored OCS button grammar.
// No inline styles or custom classes.

let myHTML = [
  '<div class="ocs__links">',
  '  <a class="ocs__btn alert-red fill pill" href="https://www.mygoodbrain.org/resources">Urgent Support</a>',
  '  <a class="ocs__btn alert-green fill pill" href="https://www.mygoodbrain.org/resources">Self-Care</a>',
  '  <a class="ocs__btn accent outline pill" href="https://www.mygoodbrain.org/resources">Community</a>',
  '</div>'
].join('');

outputElement.innerHTML =
  '<h3>Adaptive Resource Toolbar</h3>' +
  myHTML;
{% endcapture %}

{% include runners/ui.html
   runner_id="buttons-homework-1"
   challenge=homework1
   code=homework1_code
   autostart="true"
%}
```
