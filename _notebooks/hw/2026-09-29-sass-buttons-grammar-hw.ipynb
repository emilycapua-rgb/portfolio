#### Hack 2: Interactive Mindful Journal Card (Nesting & States)

**Goal:** Build a card component for a daily reflection entry that dynamically responds to focus and click interactions using SASS nesting.

1. Copy the starter SASS code block below into your notebook.
2. Complete the nested pseudo-selectors (`&:hover`, `&:focus-within`, and `&:active`).

### Homework Hack 2:

````python
%%html
<!-- UI_RUNNER: sass -->

<style lang="scss">

$card-bg: #f8fafc;
$card-border: #cbd5e1;
$active-glow: #0284c7;

.mgb-journal-card {
  background-color: $card-bg;
  border: 2px solid $card-border;
  border-radius: 12px;
  padding: 20px;
  transition: all 0.25s ease-in-out;

  // Hover: lift the card and add a soft shadow
  &:hover {
    border-color: $active-glow;
    transform: translateY(-4px);
    box-shadow: 0 8px 16px rgba(2, 132, 199, 0.12);
  }

  // Focus-within: highlight the card when the textarea is selected
  &:focus-within {
    border-color: $active-glow;
  }

  // Active: slightly press the card when clicked
  &:active {
    transform: translateY(-1px);
  }

  // Style the journal textarea
  .journal-input {
    width: 100%;
    border: 1px solid $card-border;
    border-radius: 6px;
    padding: 8px;

    // Focus: change textarea border
    &:focus {
      outline: none;
      border-color: $active-glow;
    }
  }
}

</style>

<div class="ocs__links">

  <div class="mgb-journal-card">

    <h3>Daily Mindful Journal</h3>

    <p>Write a reflection about your day.</p>

    <textarea
      class="journal-input"
      rows="4"
      placeholder="Write your reflection here...">
    </textarea>

    <br><br>

    <button class="ocs__btn accent fill">
      Save Reflection
    </button>

  </div>

</div>
````
