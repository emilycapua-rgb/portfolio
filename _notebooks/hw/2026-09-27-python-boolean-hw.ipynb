## Popcorn Hack 1: Check One SFI Part Record

An SFI automotive record is complete when it has a product name, a racing category, a spec number, and an effective date.

For this challenge, **Auto Racing** and **Drag Racing** are valid racing categories.

### Task

Use Boolean expressions to check each requirement for the SFI record below.

Before running the code, predict whether each check will be `True` or `False`.

Then change at least one value and run the code again to see how the Boolean results change.

```python
# CODE_RUNNER: Popcorn Hack 1 - Check SFI Part Fields

part = {
    "product_name": "Replacement Flywheels and Clutch Assemblies",
    "category": "Auto Racing",
    "spec_number": "1.1",
    "effective_date": "Nov. 9, 2001"
}

valid_categories = ["Auto Racing", "Drag Racing"]

# Check whether each required field is present.
has_product_name = bool(part["product_name"])
has_valid_category = part["category"] in valid_categories
has_spec_number = bool(part["spec_number"])
has_effective_date = bool(part["effective_date"])

# Print the Boolean result of each requirement.
print("Has product name:", has_product_name)
print("Has valid category:", has_valid_category)
print("Has spec number:", has_spec_number)
print("Has effective date:", has_effective_date)
```

## Popcorn Hack 2: Backend Create Rule

The backend should create a new SFI record only when:

- the product name is present,
- the category is valid,
- the spec number is present,
- the effective date is present,
- and the spec number is not already in the backend.

### Task

Build one Boolean expression that combines all of these requirements.

Run the code with the provided record.

Then change the spec number to `"2.1"` and run it again. Explain why the result changes.

```python
# CODE_RUNNER: Popcorn Hack 2 - Backend Create Rule

part = {
    "product_name": "Replacement Flywheels and Clutch Assemblies",
    "category": "Auto Racing",
    "spec_number": "1.1",
    "effective_date": "Nov. 9, 2001"
}

valid_categories = ["Auto Racing", "Drag Racing"]
existing_spec_numbers = ["1.1", "2.1"]

# Create a Boolean check for each requirement.
has_product_name = bool(part["product_name"])
valid_category = part["category"] in valid_categories
has_spec_number = bool(part["spec_number"])
has_effective_date = bool(part["effective_date"])
not_duplicate = part["spec_number"] not in existing_spec_numbers

# All requirements must be True.
can_create_record = (
    has_product_name
    and valid_category
    and has_spec_number
    and has_effective_date
    and not_duplicate
)

print("Can create SFI record:", can_create_record)
```

## Popcorn Hack 3: SFI Search Filter

An SFI backend search should match a record when either:

- the product name contains the search query, **or**
- the spec number exactly matches the search query.

The record must also belong to an accepted racing category.

### Task

Use `or` for the search match and `and` to require a valid category.

Run the code with the query `"1.2"`.

Then change the query to `"Clutch"` and run it again.

Predict which records will appear before pressing Run.

```python
# CODE_RUNNER: Popcorn Hack 3 - SFI Search Filter

query = "1.2"

records = [
    {
        "product_name": "Replacement Flywheels and Clutch Assemblies",
        "category": "Auto Racing",
        "spec_number": "1.1"
    },
    {
        "product_name": "Multiple Disc Clutch Assemblies",
        "category": "Drag Racing",
        "spec_number": "1.2"
    },
    {
        "product_name": "Racing Flywheel Record",
        "category": "Auto Racing",
        "spec_number": "2.1"
    }
]

valid_categories = ["Auto Racing", "Drag Racing"]

# Check every record.
for record in records:

    # Match either the product name or the exact spec number.
    query_matches = (
        query.lower() in record["product_name"].lower()
        or query == record["spec_number"]
    )

    # The record must also have an accepted category.
    valid_record = record["category"] in valid_categories

    if query_matches and valid_record:
        print(
            "MATCH:",
            record["product_name"],
            "| Spec:",
            record["spec_number"]
        )
```

### Homework Hack (independent): SFI Backend Validator

Build a reusable validator for candidate SFI automotive records.

A valid record has:
- a non-empty product name,
- an accepted racing category,
- a non-empty spec number,
- an effective date,
- and a spec number that does **not** already exist in the backend.

The provided test data includes different cases. Your completed runner should clearly print whether each candidate record is accepted or rejected.

Copy this starter into your **personal submission notebook** after class. Complete it there, run multiple tests, and include the output. This is a reference block, not an in-page runner:

```python
# CODE_RUNNER: Homework Hack

existing_spec_numbers = ["1.1", "2.1"]
valid_categories = ["Auto Racing", "Drag Racing"]

test_records = [
    {
        "product_name": "Multiple Disc Clutch Assemblies",
        "category": "Auto Racing",
        "spec_number": "1.2",
        "effective_date": "Feb. 9, 2006"
    },
    {
        "product_name": "Replacement Flywheels",
        "category": "Street Car",
        "spec_number": "3.1",
        "effective_date": "Jan. 1, 2026"
    },
    {
        "product_name": "Existing Flywheel Record",
        "category": "Drag Racing",
        "spec_number": "1.1",
        "effective_date": "Jan. 1, 2026"
    }
]

def validate_record(record):
    if not record["product_name"]:
        return False

    if record["category"] not in valid_categories:
        return False

    if not record["spec_number"]:
        return False

    if not record["effective_date"]:
        return False

    if record["spec_number"] in existing_spec_numbers:
        return False

    return True

for record in test_records:
    if validate_record(record):
        print(record["product_name"], "→ ACCEPTED")
    else:
        print(record["product_name"], "→ REJECTED")
```
