### Homework Hack: High School Daily Routine Dashboard

Write your own Python nested conditional from scratch to guide a student through an ordinary school day. The starter runner gives you only the student-life data; you must design the complete decision structure yourself. Your dashboard should combine account access, homework, meals, and sports practice.

Your finished program must:
1. store at least four student-life values, including login status, homework status, a meal decision, and sports-practice status,
2. use an outer account `if` / `else` before checking the student's schedule,
3. contain at least two nested decisions,
4. produce at least four clear recommendations,
5. test inputs that reach every meaningful outcome,
6. include 2–3 sentences explaining why the inner checks depend on the outer decisions.

> **Submission:** Use the runner below to build and test your dashboard. When it works, **scroll to the end of this page** to write or paste your final Python code into the Code Submission form and submit it.

```python
{% capture homework1 %}
Homework Hack: Build a High School Daily Routine Dashboard using nested conditionals. Use all four student-life values and press Run to test your decisions.
{% endcapture %}

{% capture homework1_code %}
// Starter student-life data.
// Keep these values and build your nested conditional below.

let student = {
  logged_in: true,
  homework_finished: false,
  lunch_packed: true,
  practice_today: true
};

// Write your nested conditional below.
// Use the student values to create at least four recommendations.

let recommendation;

if (student.logged_in) {

  if (student.homework_finished) {

    if (student.lunch_packed) {
      recommendation = student.practice_today
        ? "Homework finished, lunch packed, and practice is ready."
        : "Homework finished and lunch packed. No practice today.";
    } else {
      recommendation = student.practice_today
        ? "Homework finished. Buy lunch before practice."
        : "Homework finished. Buy lunch and enjoy your free afternoon.";
    }

  } else {

    if (student.lunch_packed) {
      recommendation = student.practice_today
        ? "Finish homework, then bring your packed lunch to prepare for practice."
        : "Finish homework, then use your packed lunch and free afternoon.";
    } else {
      recommendation = student.practice_today
        ? "Finish homework, buy lunch, and prepare for practice."
        : "Finish homework, then buy lunch and follow your afternoon schedule.";
    }

  }

} else {
  recommendation = "Log in first before checking your daily schedule.";
}

outputElement.innerHTML =
  '<div class="ocs__card">' +
  '<h3>High School Daily Routine Dashboard</h3>' +
  '<p><strong>Logged in:</strong> ' + student.logged_in + '</p>' +
  '<p><strong>Homework finished:</strong> ' + student.homework_finished + '</p>' +
  '<p><strong>Lunch packed:</strong> ' + student.lunch_packed + '</p>' +
  '<p><strong>Practice today:</strong> ' + student.practice_today + '</p>' +
  '<hr>' +
  '<p><strong>Recommendation:</strong> ' + recommendation + '</p>' +
  '</div>';
{% endcapture %}

{% include runners/ui.html
   runner_id="nested-conditionals-homework-1"
   challenge=homework1
   code=homework1_code
   autostart="true"
%}
```
