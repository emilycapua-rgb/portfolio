## Popcorn: Keep the Checkpoint

### Task

A player has collected 8 stars.

The game saves the current number of stars in `checkpoint_stars`. The player then earns 4 additional stars.

Predict the values printed by the program before running it.

### Questions

1. What will `stars` contain at the end?
2. What will `checkpoint_stars` contain?
3. Why are the values different?

Run the code and compare your prediction with the actual output.

```python
# CODE_RUNNER: UESL 3.1 Popcorn - Keep the Checkpoint

stars = 8

checkpoint_stars = stars

stars = stars + 4

print(stars)
print(checkpoint_stars)
```

### Checkpoint Debrief

The variable `stars` starts at 8.

The assignment `checkpoint_stars = stars` copies the current value of `stars` into `checkpoint_stars`.

When `stars = stars + 4` runs, the value of `stars` becomes 12.

The checkpoint remains 8 because it stores the value from the earlier assignment. It does not automatically update when `stars` changes.

This demonstrates the difference between a stored value and a recalculated value.

## Homework: Build Your UESL Star Counter

### Task 1: Create Your Game Profile

Choose a game name and a Boolean setting for reduced motion.

Create three variables representing the stars collected in each level:

- Level 1: 12 stars
- Level 2: 18 stars
- Level 3: 10 stars

Calculate and print the total number of stars.

Print the game name, reduced-motion setting, and total.

### Task 2: Add a Bonus

Add 5 stars to level 2.

Print the existing stored total before recalculating it.

Then recalculate the total using all three level variables and print the updated total.

### Task 3: Explain Your Results

Explain why the stored total does not change automatically when the level 2 star count changes.

### Requirements

- Use meaningful variable names.
- Use strings, integers, and a Boolean.
- Initialize every variable before using it.
- Include comments.
- Print the required values.
- Do not use `input()` or imports.

```python
# CODE_RUNNER: UESL 3.1 Homework - Build a Star Counter

# Game profile
game_name = "UESL Star Trail"
reduced_motion = True

# Starting stars for each level
level_one_stars = 12
level_two_stars = 18
level_three_stars = 10

# Calculate the original total
total_stars = level_one_stars + level_two_stars + level_three_stars

print("Game:", game_name)
print("Reduced motion:", reduced_motion)
print("Original total:", total_stars)

# Add a bonus to level 2
level_two_stars = level_two_stars + 5

# The stored total has not been recalculated yet
print("Stored total after bonus:", total_stars)

# Recalculate the total using the updated level values
total_stars = level_one_stars + level_two_stars + level_three_stars

print("Updated total:", total_stars)
```

### Homework Explanation

The original total is 40 because 12 + 18 + 10 = 40.

After adding 5 stars to level 2, that level contains 23 stars.

The variable `total_stars` still contains 40 because changing `level_two_stars` does not automatically recalculate the total.

When the program executes:

`total_stars = level_one_stars + level_two_stars + level_three_stars`

the total is recalculated as 12 + 23 + 10 = 45.

This demonstrates why a stored total must be reassigned when the values used to calculate it change.
