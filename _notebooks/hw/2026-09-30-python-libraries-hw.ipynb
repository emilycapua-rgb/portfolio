### Popcorn Hack 1: Read the API, Then Use It

Run the code below and read the documentation printed for the library procedures.

Then:

1. Change the jersey number range from `1-99` to `1-55`.
2. Change the court size from `144` to a different number.
3. Predict the new output before running the code again.
4. Run the code and compare your prediction with the result.

The goal is to practice reading an API before using a library procedure.


```python
# CODE_RUNNER: Popcorn Hack 1

import math
import random

# Read the API before calling a procedure
print("random.randint:", random.randint.__doc__)
print("math.ceil:", math.ceil.__doc__)

print("\nMath library procedures:")
print(dir(math))

print("\nRandom library procedures:")
print(dir(random))

points_per_game = [24, 18, 31, 27, 22]

# random.choice picks one item from a list
print(
    "Random game to review:",
    random.choice(points_per_game),
    "points"
)

# Changed jersey range from 1-99 to 1-55
jersey = random.randint(1, 55)
print("Random jersey number:", jersey)

# Changed court size from 144 to 196
side = math.sqrt(196)
print("Square court side:", side, "feet")
```

### Popcorn Hack 2: Write a Flask Route

Run the code to see both Flask routes answer.

Then:

1. Change the players in the roster.
2. Add a route of your own, such as `/coach` or `/record`.
3. Remember that routes must be defined **above** `app.test_client()`.
4. Run the program and check that your new route returns a response.


```python
# CODE_RUNNER: Popcorn Hack 2

from flask import Flask, jsonify

app = Flask(__name__)

# Existing score route
@app.route("/score")
def score():
    return jsonify({
        "home": 58,
        "away": 52,
        "quarter": 3
    })


# Changed the roster
@app.route("/roster")
def roster():
    return jsonify({
        "players": ["Emily", "Jordan", "Avery"]
    })


# Added my own route
@app.route("/coach")
def coach():
    return jsonify({
        "coach": "Coach Taylor",
        "message": "Keep practicing!"
    })


# test_client visits the routes without a browser
client = app.test_client()

print("/score says:", client.get("/score").get_json())
print("/roster says:", client.get("/roster").get_json())
print("/coach says:", client.get("/coach").get_json())
```

### Homework Hack: Season Stats Report

The cell below is a **worked example**: a finished program that reports on a player's season using three libraries. Run it, read how each library is used, then build your own version in your homework notebook.

**Your version must:**

1. **Use at least 3 standard library modules.** This one uses `json`, `datetime`, and `statistics`. Swap in or add others, such as `random` or `math`.
2. **Have a procedure like `summarize_season`** that returns the total points, average points, best game, and points against each opponent.
3. **Document your API.** Write a docstring giving the parameters and the return value, so someone else could use your procedure without reading the code.
4. **Use `datetime`** to report how many days are left until your next game.
5. **Write a `requirements.txt`.** Imagine a real version of this app that uses `pandas` for the season table and `Flask` for the scoreboard page. List those packages, and add a comment explaining why `json` and `datetime` are not listed.

Use your own sport, your own games, and your own next game date, not the sample values. Any sport works: swap points for goals, times, or kills.

```python
# CODE_RUNNER: Homework Hack

import json
import statistics
from datetime import date


# My own season data
games = [
    {
        "date": "2026-09-14",
        "opponent": "Lincoln",
        "points": 2,
        "minutes": 70
    },
    {
        "date": "2026-09-18",
        "opponent": "Madison",
        "points": 1,
        "minutes": 65
    },
    {
        "date": "2026-09-22",
        "opponent": "Lincoln",
        "points": 3,
        "minutes": 75
    },
    {
        "date": "2026-09-27",
        "opponent": "Roosevelt",
        "points": 2,
        "minutes": 68
    }
]


def summarize_season(season):
    """
    Summarize a list of games.

    Parameters:
        season: list of dictionaries containing date, opponent,
                points, and minutes.

    Returns:
        A dictionary containing total_points, average_points,
        best_game, and points_by_opponent.
    """

    points = []

    for game in season:
        points.append(game["points"])

    total_points = sum(points)
    average_points = statistics.mean(points)

    best_game = max(
        season,
        key=lambda game: game["points"]
    )

    points_by_opponent = {}

    for game in season:
        opponent = game["opponent"]

        if opponent not in points_by_opponent:
            points_by_opponent[opponent] = 0

        points_by_opponent[opponent] += game["points"]

    return {
        "total_points": total_points,
        "average_points": average_points,
        "best_game": best_game,
        "points_by_opponent": points_by_opponent
    }


# Create the season report
report = summarize_season(games)

print("SEASON STATS REPORT")
print(json.dumps(report, indent=2))


# Calculate days until the next game
today = date(2026, 9, 30)
next_game = date(2026, 10, 5)

days_until_next_game = (next_game - today).days

print()
print("Next game:", next_game)
print("Days until next game:", days_until_next_game)


# requirements.txt for a real version of the project
requirements = """
pandas
Flask

# json and datetime are not listed because
# they are part of Python's standard library.
"""

print()
print("requirements.txt:")
print(requirements)
```
