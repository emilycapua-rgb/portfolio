## Popcorn Hack A: Find the Book

A librarian needs to find a book in a catalog.

Linear search checks books one at a time. Binary search repeatedly
halves a sorted catalog to narrow down where the book is located.

### Your tasks
1. Predict the output before running the code.
2. Run the code and compare the number of checks.
3. Change the counts to model searching for book 3:
   - Linear search takes 1 check.
   - Binary search takes 3 checks.
4. Explain which method uses fewer checks in each example.

**Question:** Why can two correct algorithms require different amounts of work?

```python
# CODE_RUNNER: Algorithmic Efficiency Popcorn A - Find the Book

# Searching for book 15
linear_checks = 5
binary_checks = 3

print("Searching for book 15")
print("Linear search checks:", linear_checks)
print("Binary search checks:", binary_checks)
print("Checks saved with binary search:", linear_checks - binary_checks)

# Searching for book 3
print("\nSearching for book 3")
linear_checks = 1
binary_checks = 3

print("Linear search checks:", linear_checks)
print("Binary search checks:", binary_checks)
print("Checks saved with binary search:", linear_checks - binary_checks)
```

## Popcorn Hack B: Expand the Catalog

A librarian wants to estimate how many entries two search methods
might inspect in the worst case.

Linear search may inspect every entry. Binary search repeatedly
halves a sorted catalog.

### Your tasks
1. Predict the number of checks for a catalog of 8 books.
2. Run the code.
3. Change `n` from 8 to 16.
4. Compare how the number of checks changes.
5. Explain why binary search requires a sorted catalog and direct
   access to middle entries.

This calculator estimates the number of inspections; it does not
perform an actual search.

```python
# CODE_RUNNER: Algorithmic Efficiency Popcorn B - Expand the Catalog

import math

n = 8

linear_checks = n
binary_checks = 0 if n == 0 else math.floor(math.log2(n)) + 1

print("Catalog size:", n)
print("Linear search worst-case checks:", linear_checks)
print("Binary search estimated worst-case checks:", binary_checks)

print("\nRecommendation:")
if n > 0:
    print("For repeated searches, consider binary search if the catalog is sorted.")
else:
    print("The catalog is empty, so no book entries need to be inspected.")
```
