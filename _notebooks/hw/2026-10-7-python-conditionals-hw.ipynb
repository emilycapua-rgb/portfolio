## Popcorn Hack 2: Fix the elif Chain

### Task

The original code checks `risk_score >= 3` before checking `risk_score >= 6` and `risk_score >= 9`.

This causes a problem because any score greater than or equal to 9 also satisfies the first condition.

As a result, some branches can never run.

### Requirements

1. Order the conditions from highest threshold to lowest.
2. Keep all four possible outcomes.
3. Add a count comment.
4. Make sure every branch can be reached.
5. Run the code and check the output.

```python
# CODE_RUNNER: 3.6 Popcorn 2 - Fix the elif Chain

risk_score = 9

# 3 thresholds -> 4 outcomes -> 4 answers

if risk_score >= 9:
    print("Do not reply. This is almost certainly a scam.")

elif risk_score >= 6:
    print("This looks risky. Check with someone you trust before replying.")

elif risk_score >= 3:
    print("Be a little careful, but this is probably fine.")

else:
    print("Nothing suspicious in this message.")
```

### Why Was the Original Code Incorrect?

The original code checked `risk_score >= 3` first.

When `risk_score` was 9, that condition was already True, so Python skipped the remaining `elif` statements.

The branches checking `risk_score >= 6` and `risk_score >= 9` could never run.

The fix is to check the highest threshold first, followed by the lower thresholds.

This makes all four branches reachable.

## Homework Hack: Score a Message, Then Give Advice

### Part 1: Calculate the Risk Score

Start with `risk_score = 0`.

Use a separate `if` statement for each signal:

- Add 4 points if the message contains `"gift card"`.

### Part 2: Give Advice

Use an `if` / `elif` / `else` chain with four possible outcomes:

- A score of 7 or higher: Almost certainly a scam.
- A score of 4–6: Risky; check with someone you trust.
- A score of 2–3: Be careful.
- A score of 0–1: No warning signals detected by this simple trainer.

Order the thresholds from highest to lowest.

### Part 3: Prove Every Branch Is Reachable

### Part 4: Test the Extremes

Test both:

```python
# CODE_RUNNER: 3.6 Homework - Score a Message and Give Advice

# ==========================================
# TEST 1: CLEAN MESSAGE
# ==========================================

message = "Hi, are we still meeting on Sunday?"
amount_requested = 0
known_contact = True

risk_score = 0

# No else is needed because a missing signal adds zero points.
if "gift card" in message:
    risk_score += 4

# No else is needed because an amount of 100 or less adds zero points.
if amount_requested > 100:
    risk_score += 3

# No else is needed because a known contact adds zero points.
if not known_contact:
    risk_score += 2

print("CLEAN MESSAGE")
print("Risk score:", risk_score)

# 3 thresholds -> 4 outcomes -> 4 answers

if risk_score >= 7:
    print("Advice: Do not reply. This is almost certainly a scam.")

elif risk_score >= 4:
    print("Advice: This looks risky. Check with someone you trust before replying.")

elif risk_score >= 2:
    print("Advice: Be a little careful, but this may be okay.")

else:
    print("Advice: No warning signals were detected by this simple trainer.")

print()

# ==========================================
# TEST 2: ALL WARNING SIGNALS
# ==========================================

message = "URGENT: send $500 in gift cards now to unlock your account"
amount_requested = 500
known_contact = False

risk_score = 0

if "gift card" in message:
    risk_score += 4

if amount_requested > 100:
    risk_score += 3

if not known_contact:
    risk_score += 2

print("ALL-SIGNALS MESSAGE")
print("Risk score:", risk_score)

# 3 thresholds -> 4 outcomes -> 4 answers

if risk_score >= 7:
    print("Advice: Do not reply. This is almost certainly a scam.")

elif risk_score >= 4:
    print("Advice: This looks risky. Check with someone you trust before replying.")

elif risk_score >= 2:
    print("Advice: Be a little careful, but this may be okay.")

else:
    print("Advice: No warning signals were detected by this simple trainer.")
```
