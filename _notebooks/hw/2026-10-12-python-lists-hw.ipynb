## Popcorn Hack 2: Volunteer Hours

Shoreline Community Services tracks the hours contributed by volunteers.

### Tasks
1. Use a `for` loop to calculate the total volunteer hours.
2. Calculate the average using the total and the length of the list.
3. Print both results.
4. Change the list values and test your code again.
5. Consider what should happen if the list is empty.

Remember:
- Initialize `total = 0` before the loop.
- Add each value to `total` inside the loop.
- Check that the list is not empty before calculating the average.
- Do not use `sum()`.

```python
# CODE_RUNNER: Lists Popcorn 2 - Volunteer Hours

volunteer_hours = [7, 6, 8, 5, 7, 9]

total = 0

for hours in volunteer_hours:
    total += hours

print("Total volunteer hours:", total)

if len(volunteer_hours) > 0:
    average = total / len(volunteer_hours)
    print("Average volunteer hours:", average)
else:
    print("No data yet")
```

## Popcorn Hack 3: Counting and Filtering Donations

Shoreline Community Services wants to find out which supplies are
requested most often.

### Tasks
1. Count how many times each item appears using a dictionary.
2. Create an empty list called `popular`.
3. Loop through the frequency dictionary.
4. Add items with two or more requests to `popular`.
5. Print the frequency dictionary and the popular items.

Remember:
- Start with an empty dictionary: `frequency = {}`.
- Check whether each item already exists in the dictionary.
- Start the filtered list with `popular = []`.
- Use `.append()` to add qualifying items.
- Do not use list comprehensions or modify the original list.

```python
# CODE_RUNNER: Lists Popcorn 3 - Count and Filter

items_requested = [
    "blankets", "socks", "meals", "socks",
    "water", "socks", "meals"
]

frequency = {}

for item in items_requested:
    if item in frequency:
        frequency[item] += 1
    else:
        frequency[item] = 1

print("Donation frequencies:", frequency)

popular = []

for item in frequency:
    if frequency[item] >= 2:
        popular.append(item)

print("Items requested two or more times:", popular)
```

## Homework Hack: Donation Supply Analysis

Shoreline Community Services is analyzing donated supplies and their
associated sourcing costs.

Use lists, indexing, accumulator loops, dictionaries, and filtering
to analyze the data.

### Part 1: Indexing
1. Print the number of donations.
2. Print the last donation using indexing.
3. Do not hardcode the last index.

### Part 2: Sum and Average
4. Use a `for` loop to calculate the total cost.
5. Calculate the average cost.
6. Print the total and average as integers using `int()`.

### Part 3: Frequency Counting
7. Build a dictionary counting each donation item.
8. Print the dictionary.

### Part 4: Filtering
9. Create a new list containing items donated three or more times.
10. Print the filtered list.

### Requirements
- Use loops instead of `sum()` or list comprehensions.
- Initialize the total before the loop.
- Initialize the dictionary and filtered list before using them.
- Keep the original donation list unchanged.
- Run the code and check the output against the expected results.

```python
# CODE_RUNNER: Lists Homework - Donation Supply Analysis

donations_by_item = [
    "blankets", "socks", "blankets", "water",
    "socks", "blankets", "meals", "socks",
    "blankets", "water", "blankets", "socks"
]

costs = [2, 3, 2, 4, 3, 2, 1, 3, 2, 4, 2, 3]

# LESSON 1 - Indexing

print("Number of donations:", len(donations_by_item))
print("Last donation:", donations_by_item[-1])

# LESSON 2 - Sum and Average

total = 0

for cost in costs:
    total += cost

print("Total:", int(total))

if len(costs) > 0:
    average = total / len(costs)
    print("Average:", int(average))
else:
    print("Average: No data yet")

# LESSON 3 - Frequency Counting

frequency = {}

for item in donations_by_item:
    if item in frequency:
        frequency[item] += 1
    else:
        frequency[item] = 1

print("Donation frequencies:", frequency)

# LESSON 3 - Filtering

popular = []

for item in frequency:
    if frequency[item] >= 3:
        popular.append(item)

print("Items donated three or more times:", popular)
```
