## Task

Refactor the following code block.

1. Remove all custom CSS and classes.
2. Replace them with the correct semantic HTML tags:
   - `h2`–`h4`
   - `p`
   - `ol` / `li`
   - `strong`
3. Apply the correct OCS classes, such as:
   - `ocs_card`
   - `ocs_section-title`
4. Make sure the code uses the **Aesthetihawk SASS theme**.
5. Run the result with `UI_RUNNER`.
6. Submit the clean HTML in your notebook.
